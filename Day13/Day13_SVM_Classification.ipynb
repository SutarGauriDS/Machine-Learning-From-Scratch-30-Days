# Day 13 — Support Vector Machine (SVM)

Classification using the built-in Breast Cancer dataset from scikit-learn.

## 1. Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
```

## 2. Load the Dataset

```python
data = load_breast_cancer()

X = pd.DataFrame(
    data.data,
    columns=data.feature_names
)

y = pd.Series(
    data.target,
    name="target"
)

print("Dataset Shape:", X.shape)
print("\nTarget Classes:", data.target_names)
print("\nFirst 5 Rows:")
display(X.head())
```

## 3. Check Class Distribution

```python
print("Class Distribution:")
print(y.value_counts())

plt.figure(figsize=(6, 4))
y.value_counts().sort_index().plot(kind="bar")
plt.xticks(
    ticks=[0, 1],
    labels=data.target_names,
    rotation=0
)
plt.xlabel("Class")
plt.ylabel("Number of Samples")
plt.title("Target Class Distribution")
plt.tight_layout()
plt.show()
```

## 4. Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Samples:", X_train.shape[0])
print("Testing Samples :", X_test.shape[0])
```

## 5. Feature Scaling

SVM is sensitive to feature scale, so StandardScaler is applied using only the training data.

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed.")
```

## 6. Linear SVM

```python
linear_svm = SVC(
    kernel="linear",
    C=1
)

linear_svm.fit(X_train_scaled, y_train)

linear_pred = linear_svm.predict(X_test_scaled)

linear_accuracy = accuracy_score(
    y_test,
    linear_pred
)

print(f"Linear SVM Accuracy: {linear_accuracy:.4f}")
```

## 7. RBF / Non-Linear SVM

```python
rbf_svm = SVC(
    kernel="rbf",
    C=1,
    gamma="scale"
)

rbf_svm.fit(X_train_scaled, y_train)

rbf_pred = rbf_svm.predict(X_test_scaled)

rbf_accuracy = accuracy_score(
    y_test,
    rbf_pred
)

print(f"RBF SVM Accuracy: {rbf_accuracy:.4f}")
```

## 8. Compare Linear and RBF SVM

```python
comparison = pd.DataFrame({
    "Model": ["Linear SVM", "RBF SVM"],
    "Accuracy": [linear_accuracy, rbf_accuracy]
})

display(comparison)

if rbf_accuracy >= linear_accuracy:
    best_predictions = rbf_pred
    best_name = "RBF SVM"
    best_accuracy = rbf_accuracy
else:
    best_predictions = linear_pred
    best_name = "Linear SVM"
    best_accuracy = linear_accuracy

print(f"Best Model: {best_name}")
print(f"Best Accuracy: {best_accuracy:.4f}")
```

## 9. Accuracy Comparison

```python
plt.figure(figsize=(7, 5))
plt.bar(comparison["Model"], comparison["Accuracy"])
plt.ylim(0.8, 1.0)
plt.ylabel("Accuracy")
plt.xlabel("SVM Model")
plt.title("Linear SVM vs RBF SVM")
plt.tight_layout()
plt.show()
```

## 10. Classification Report

```python
print(classification_report(
    y_test,
    best_predictions,
    target_names=data.target_names
))
```

## 11. Confusion Matrix

```python
cm = confusion_matrix(y_test, best_predictions)

print("Confusion Matrix:")
print(cm)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=data.target_names
)

disp.plot()
plt.title(f"{best_name} - Confusion Matrix")
plt.tight_layout()
plt.show()
```

## 12. Final Result

This notebook compares a Linear SVM with an RBF (non-linear) SVM after feature scaling.

```python
print("=" * 50)
print("FINAL RESULT")
print("=" * 50)
print(f"Linear SVM Accuracy : {linear_accuracy:.4f}")
print(f"RBF SVM Accuracy    : {rbf_accuracy:.4f}")
print(f"Best Model          : {best_name}")
print(f"Best Accuracy       : {best_accuracy:.4f}")
print("\nDay 13 SVM analysis complete!")
```
