## 📚 What is NumPy?

NumPy stands for Numerical Python.

It is a Python library used for:
- Numerical calculations
- Working with arrays
- Mathematical operations
- Statistical operations
- Handling numerical data efficiently

NumPy arrays are widely used in Data Science and Machine Learning.

```python
import numpy as np

print("NumPy Version:", np.__version__)
```

```python
arr = np.array([10, 20, 30, 40, 50])

print("Array:")
print(arr)
```

```python
python_list = [10, 20, 30, 40, 50]

numpy_array = np.array([10, 20, 30, 40, 50])

print("Python List:", python_list)
print("NumPy Array:", numpy_array)
```

```python
arr_1d = np.array([10, 20, 30, 40, 50])

print(arr_1d)
print("Dimensions:", arr_1d.ndim)
```

```python
arr_2d = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(arr_2d)
print("Dimensions:", arr_2d.ndim)
```

```python
print("Shape:", arr_2d.shape)
```

```python
print("Total Elements:", arr_2d.size)
```

```python
print("Data Type:", arr_2d.dtype)
```

```python
arr = np.array([10, 20, 30, 40, 50])

print("First Element:", arr[0])
print("Third Element:", arr[2])
print("Last Element:", arr[-1])
```

```python
print(arr[1:4])
```

```python
arr = np.array([10, 20, 30, 40])

print("Addition:", arr + 5)
print("Subtraction:", arr - 5)
print("Multiplication:", arr * 2)
print("Division:", arr / 2)
```

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
```

```python
marks = np.array([60, 70, 80, 90, 100])

print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("Standard Deviation:", np.std(marks))
```

```python
print("Total:", np.sum(marks))
```

```python
arr = np.array([1, 2, 3, 4, 5, 6])

reshaped_arr = arr.reshape(2, 3)

print("Original Array:")
print(arr)

print("\nReshaped Array:")
print(reshaped_arr)
```

```python
zeros = np.zeros((2, 3))
ones = np.ones((2, 3))

print("Zeros:")
print(zeros)

print("\nOnes:")
print(ones)
```

```python
numbers = np.arange(1, 11)

print(numbers)
```

```python
np.random.seed(42)

random_numbers = np.random.randint(1, 100, 5)

print(random_numbers)
```

## 🤖 Why is NumPy Important in Machine Learning?

NumPy is useful for:

- Representing numerical data as arrays
- Performing mathematical calculations
- Working with matrices
- Preparing numerical features
- Performing statistical calculations
- Supporting libraries such as Pandas and Scikit-Learn

Machine Learning involves a lot of numerical calculations, so understanding NumPy is an important foundation.

```python
# Create marks for 5 students

marks = np.array([72, 85, 91, 68, 79])

print("Marks:", marks)
print("Average:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
print("Standard Deviation:", np.std(marks))
```
