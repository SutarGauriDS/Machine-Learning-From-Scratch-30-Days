# 🚀 Day 23 — Cross Validation

## Model Performance Evaluation using K-Fold Cross Validation

Part of my **30 Days of Machine Learning Challenge**.

## 🎯 Objective

Evaluate and compare Logistic Regression, KNN, and Random Forest using **5-Fold Stratified Cross-Validation**.

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
```

## 1. Load Dataset

```python
data = load_breast_cancer()
X = pd.DataFrame(data.data, columns=data.feature_names)
y = pd.Series(data.target, name='target')
print('Dataset Shape:', X.shape)
print('Target Classes:', data.target_names)
X.head()
```

## 2. Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
print('Training Samples:', X_train.shape[0])
print('Testing Samples :', X_test.shape[0])
```

## 3. Define Models

Scaling is applied inside pipelines for Logistic Regression and KNN. Random Forest does not require scaling.

```python
models = {
    'Logistic Regression': Pipeline([('scaler', StandardScaler()), ('model', LogisticRegression(max_iter=5000))]),
    'KNN': Pipeline([('scaler', StandardScaler()), ('model', KNeighborsClassifier(n_neighbors=5))]),
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42)
}
models
```

## 4. Define 5-Fold Stratified Cross-Validation

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
print('Number of folds:', cv.n_splits)
```

## 5. Perform Cross Validation

```python
results = []
for model_name, model in models.items():
    scores = cross_val_score(model, X_train, y_train, cv=cv, scoring='accuracy', n_jobs=-1)
    results.append({'Model': model_name, 'Mean Accuracy': scores.mean(), 'Std Deviation': scores.std()})
    print(f'\n{model_name}')
    print('Fold Scores:', scores)
    print(f'Mean Accuracy: {scores.mean():.4f}')
    print(f'Std Deviation: {scores.std():.4f}')
```

## 6. Compare Model Performance

```python
results_df = pd.DataFrame(results).sort_values('Mean Accuracy', ascending=False)
results_df
```

### 📌 Interpretation

- Higher **Mean Accuracy** indicates better average validation performance.
- Lower **Standard Deviation** indicates more consistent performance across folds.

## 7. Train the Best Model and Evaluate on Test Data

```python
best_model_name = results_df.iloc[0]['Model']
best_model = models[best_model_name]
best_model.fit(X_train, y_train)
test_accuracy = best_model.score(X_test, y_test)
print('Best Model:', best_model_name)
print(f'Mean CV Accuracy: {results_df.iloc[0]["Mean Accuracy"]:.4f}')
print(f'Test Accuracy: {test_accuracy:.4f}')
```

## 8. Save Results

```python
results_df.to_csv('Day23_Cross_Validation_Results.csv', index=False)
print('Results saved to Day23_Cross_Validation_Results.csv')
```

## 💡 Key Learnings

1. Cross-validation gives a more reliable estimate of model performance.
2. Stratified K-Fold preserves class proportions in classification tasks.
3. Mean CV accuracy measures average performance.
4. Standard deviation helps understand model stability.
5. The test set should remain unseen until final evaluation.
