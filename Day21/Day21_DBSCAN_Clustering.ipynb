# 🚀 Day 21 — DBSCAN Clustering

## Customer Segmentation using Density-Based Clustering

Part of my **30 Days of Machine Learning Challenge**.

### 🎯 Objective
Use DBSCAN to group customers based on Annual Income and Spending Score, while identifying noise/outliers.

```
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
```

## 1. Load Dataset

```
df = pd.read_csv("Mall_Customers.csv")
print("Dataset Shape:", df.shape)
display(df.head())
```

## 2. Explore Dataset

```
print("Columns:")
print(df.columns.tolist())
print("\nMissing Values:")
print(df.isnull().sum())
df.info()
```

## 3. Feature Selection

```
features = ["Annual Income (k$)", "Spending Score (1-100)"]
missing = [c for c in features if c not in df.columns]
if missing: raise ValueError(f"Required columns are missing: {missing}")
X = df[features]
display(X.head())
```

## 4. Original Data

```
plt.figure(figsize=(9,6))
plt.scatter(X["Annual Income (k$)"], X["Spending Score (1-100)"], alpha=0.7)
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.title("Customer Data Before DBSCAN")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## 5. Feature Scaling

```
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print("Features standardized successfully.")
```

## 6. Apply DBSCAN

`eps` controls neighborhood distance and `min_samples` controls the minimum points needed for a dense region.

```
eps_value = 0.35
min_samples_value = 5
dbscan = DBSCAN(eps=eps_value, min_samples=min_samples_value)
cluster_labels = dbscan.fit_predict(X_scaled)
print("eps:", eps_value)
print("min_samples:", min_samples_value)
print("DBSCAN clustering completed.")
```

## 7. Add Cluster Labels

```
df["Cluster"] = cluster_labels
display_columns = ["CustomerID", "Age", "Annual Income (k$)", "Spending Score (1-100)", "Cluster"]
display_columns = [c for c in display_columns if c in df.columns]
display(df[display_columns].head(15))
```

## 8. Identify Clusters and Noise

```
unique_labels = set(cluster_labels)
cluster_labels_without_noise = [x for x in unique_labels if x != -1]
number_of_clusters = len(cluster_labels_without_noise)
noise_count = list(cluster_labels).count(-1)
print("Number of Clusters:", number_of_clusters)
print("Number of Noise Points:", noise_count)
```

## 9. Cluster Counts

```
cluster_counts = df["Cluster"].value_counts().sort_index()
display(cluster_counts.to_frame("Customer Count"))
```

## 10. Visualize DBSCAN Clusters

```
plt.figure(figsize=(10,7))
for cluster in sorted(df["Cluster"].unique()):
    data = df[df["Cluster"] == cluster]
    if cluster == -1:
        plt.scatter(data["Annual Income (k$)"], data["Spending Score (1-100)"], label="Noise / Outliers", marker="x", s=80)
    else:
        plt.scatter(data["Annual Income (k$)"], data["Spending Score (1-100)"], label=f"Cluster {cluster}", alpha=0.7)
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.title("DBSCAN Customer Segmentation")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## 11. Cluster Summary

```
clustered = df[df["Cluster"] != -1]
if not clustered.empty:
    summary = clustered.groupby("Cluster")[["Age", "Annual Income (k$)", "Spending Score (1-100)"]].mean().round(2)
    display(summary)
else:
    print("No clusters detected with the current parameters.")
```

## 12. Noise / Outlier Analysis

```
noise_points = df[df["Cluster"] == -1]
print(f"Noise points detected: {len(noise_points)}")
if not noise_points.empty: display(noise_points[display_columns])
```

## 13. DBSCAN vs K-Means

```
comparison = pd.DataFrame({"K-Means":["Requires K","Centroid-based","Sensitive to outliers","Compact clusters"],"DBSCAN":["Does not require K","Density-based","Can identify noise","Irregular clusters"]}, index=["Number of Clusters","Approach","Outliers","Cluster Shape"])
display(comparison)
```

## 14. Save Results

```
output_file = "Mall_Customers_DBSCAN_Clustered.csv"
df.to_csv(output_file, index=False)
print(f"Saved: {output_file}")
```

## 🏁 Key Learnings

- DBSCAN is an unsupervised clustering algorithm.
- It does not require the number of clusters beforehand.
- `eps` and `min_samples` control cluster formation.
- DBSCAN can identify noise/outliers.
- Feature scaling is important for distance-based clustering.
