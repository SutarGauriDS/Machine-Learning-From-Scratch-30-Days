Use **GridSearchCV** to find better hyperparameters for a Random Forest Classifier and compare the tuned model with a baseline model.

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, ConfusionMatrixDisplay
```

## 1. Load the Dataset

```python
data = load_breast_cancer()

X = pd.DataFrame(data.data, columns=data.feature_names)
y = pd.Series(data.target, name='target')

print('Dataset Shape:', X.shape)
display(X.head())
```

## 2. Target Information

```python
print('Target Classes:')
print(data.target_names)

print('\nTarget Distribution:')
print(y.value_counts())
```

## 3. Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print('Training Samples:', len(X_train))
print('Testing Samples :', len(X_test))
```

## 4. Baseline Random Forest

```python
baseline_model = RandomForestClassifier(random_state=42)
baseline_model.fit(X_train, y_train)
baseline_predictions = baseline_model.predict(X_test)
baseline_accuracy = accuracy_score(y_test, baseline_predictions)

print(f'Baseline Accuracy: {baseline_accuracy:.4f}')
```

## 5. Define Hyperparameter Grid

```python
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 5, 10],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2],
    'max_features': ['sqrt', 'log2']
}

for parameter, values in param_grid.items():
    print(f'{parameter}: {values}')
```

## 6. Run GridSearchCV

GridSearchCV tests combinations of hyperparameters using **5-fold cross-validation**.

```python
rf_model = RandomForestClassifier(random_state=42)

grid_search = GridSearchCV(
    estimator=rf_model,
    param_grid=param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1,
    verbose=1
)

grid_search.fit(X_train, y_train)
print('Grid search completed successfully.')
```

## 7. Best Hyperparameters

```python
print('Best Hyperparameters:')
for parameter, value in grid_search.best_params_.items():
    print(f'{parameter}: {value}')
```

## 8. Best Cross-Validation Score

```python
best_cv_score = grid_search.best_score_
print(f'Best CV Accuracy: {best_cv_score:.4f}')
```

## 9. Evaluate the Tuned Model

```python
best_model = grid_search.best_estimator_
tuned_predictions = best_model.predict(X_test)
tuned_accuracy = accuracy_score(y_test, tuned_predictions)

print(f'Tuned Model Accuracy: {tuned_accuracy:.4f}')
```

## 10. Baseline vs Tuned Model

```python
comparison = pd.DataFrame({
    'Model': ['Baseline Random Forest', 'Tuned Random Forest'],
    'Accuracy': [baseline_accuracy, tuned_accuracy]
})
comparison['Accuracy'] = comparison['Accuracy'].round(4)
display(comparison)

improvement = tuned_accuracy - baseline_accuracy
print(f'Accuracy Improvement: {improvement:.4f}')
```

## 11. Classification Report

```python
print(classification_report(
    y_test,
    tuned_predictions,
    target_names=data.target_names
))
```

## 12. Confusion Matrix

```python
cm = confusion_matrix(y_test, tuned_predictions)
print(cm)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=data.target_names
)
disp.plot()
plt.title('Tuned Random Forest - Confusion Matrix')
plt.tight_layout()
plt.show()
```

## 13. Feature Importance

```python
feature_importance = pd.DataFrame({
    'Feature': X.columns,
    'Importance': best_model.feature_importances_
}).sort_values('Importance', ascending=False)

display(feature_importance.head(10))
```

## 14. Top Feature Importances

```python
top_features = feature_importance.head(10)

plt.figure(figsize=(10, 6))
plt.barh(top_features['Feature'][::-1], top_features['Importance'][::-1])
plt.xlabel('Importance')
plt.ylabel('Feature')
plt.title('Top 10 Random Forest Feature Importances')
plt.tight_layout()
plt.show()
```

## 🏁 Final Results

The project used **GridSearchCV with 5-fold cross-validation** to tune a Random Forest Classifier.

### Key Learnings

- Hyperparameters control how a machine learning model behaves.
- GridSearchCV systematically tests parameter combinations.
- Cross-validation helps evaluate configurations during model selection.
- `best_params_` returns the best parameter combination.
- `best_estimator_` returns the best fitted model.
- Comparing baseline and tuned performance shows whether tuning helped.
