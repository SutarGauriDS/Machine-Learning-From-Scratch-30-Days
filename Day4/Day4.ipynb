## 📚 What is Pandas?

Pandas is a Python library used for:

- Data manipulation
- Data cleaning
- Data analysis
- Working with structured/tabular data

The two main Pandas structures are:

1. Series
2. DataFrame

```python
import pandas as pd

print("Pandas Version:", pd.__version__)
```

```python
marks = pd.Series([70, 80, 90, 85, 95])

print(marks)
```

```python
data = {
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Neha"],
    "Age": [22, 21, 23, 22, 24],
    "City": ["Pune", "Mumbai", "Pune", "Nashik", "Mumbai"],
    "Marks": [85, 90, 78, 88, 95]
}

df = pd.DataFrame(data)

df
```

```python
print(df.head())
```

```python
print(df.tail())
```

```python
print("Shape:", df.shape)
```

```python
print("Columns:")
print(df.columns)
```

```python
df.info()
df.describe()
```

```python
print(df["Marks"])
print(df[["Name", "Marks"]])
```

```python
print(df.iloc[0:3])
```

```python
high_marks = df[df["Marks"] > 80]

print(high_marks)
```

```python
pune_students = df[df["City"] == "Pune"]

print(pune_students)
```

```python
result = df[
    (df["Marks"] > 80) &
    (df["Age"] > 21)
]

print(result)
```

```python
sorted_df = df.sort_values("Marks")

print(sorted_df)

sorted_df = df.sort_values("Marks", ascending=False)

print(sorted_df)
```

```python
df_missing = df.copy()

df_missing.loc[2, "Marks"] = None

print(df_missing)
```

```python
print(df_missing.isnull().sum())
```

```python
df_missing["Marks"] = df_missing["Marks"].fillna(
    df_missing["Marks"].mean()
)

print(df_missing)
```

```python
city_average = df.groupby("City")["Marks"].mean()

print(city_average)
```

```python
city_count = df.groupby("City")["Name"].count()

print(city_count)
```

```python
print("Average Marks:", df["Marks"].mean())
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())
```

```python
df["Result"] = df["Marks"].apply(
    lambda x: "Pass" if x >= 40 else "Fail"
)

df
```

## 🤖 Why is Pandas Important in Machine Learning?

Pandas is commonly used before model training to:

- Load datasets
- Explore data
- Clean data
- Handle missing values
- Filter records
- Create features
- Analyze patterns
- Prepare data for Machine Learning

Pandas and NumPy are often used together in ML workflows.

```python
students = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E"],
    "Study_Hours": [2, 4, 5, 3, 6],
    "Marks": [55, 70, 82, 65, 90]
})

print("Dataset:")
print(students)

print("\nAverage Marks:")
print(students["Marks"].mean())

print("\nStudents scoring above 70:")
print(students[students["Marks"] > 70])

print("\nHighest Marks:")
print(students["Marks"].max())
```
