# 🚀 Day 20 — Hierarchical Clustering

## Customer Segmentation using Agglomerative Hierarchical Clustering

**Dataset:** Mall Customers

**Features:** Annual Income and Spending Score

### Objective
Group similar customers into meaningful segments using unsupervised learning and understand how hierarchical clustering differs from K-Means.


## 1. Learning Objectives

- Understand hierarchical clustering
- Learn agglomerative clustering
- Understand linkage and dendrograms
- Scale features before clustering
- Select a suitable number of clusters
- Analyze and visualize customer segments


```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage
```

## 2. Load the Dataset

```python
df = pd.read_csv("Mall_Customers.csv")

print("Dataset Shape:", df.shape)
display(df.head())
```

## 3. Explore the Dataset

```python
print("Columns:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDataset Information:")
df.info()
```

## 4. Feature Selection

For customer segmentation, we use **Annual Income (k$)** and **Spending Score (1-100)**.


```python
features = [
    "Annual Income (k$)",
    "Spending Score (1-100)"
]

missing_columns = [col for col in features if col not in df.columns]

if missing_columns:
    raise ValueError(f"Required columns are missing: {missing_columns}")

X = df[features]

display(X.head())
```

## 5. Visualize the Original Customer Data

```python
plt.figure(figsize=(9, 6))

plt.scatter(
    X["Annual Income (k$)"],
    X["Spending Score (1-100)"],
    alpha=0.7
)

plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.title("Customer Data Before Clustering")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## 6. Feature Scaling

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Features standardized successfully.")
print("\nFirst 5 scaled rows:")
print(X_scaled[:5])
```

## 7. Create the Linkage Matrix

The **Ward linkage** method merges clusters while minimizing the increase in within-cluster variance.


```python
linked = linkage(X_scaled, method="ward")

print("Linkage matrix created successfully.")
print("\nFirst 5 rows:")
print(linked[:5])
```

## 8. Dendrogram

```python
plt.figure(figsize=(14, 7))

dendrogram(
    linked,
    truncate_mode="lastp",
    p=30,
    leaf_rotation=90,
    leaf_font_size=9
)

plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Cluster / Data Points")
plt.ylabel("Distance")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### Choosing the Number of Clusters

The dendrogram helps us inspect where large vertical gaps occur. For this project, we use **5 clusters** to maintain consistency with the previous K-Means customer segmentation project.


## 9. Apply Agglomerative Clustering

```python
n_clusters = 5

model = AgglomerativeClustering(
    n_clusters=n_clusters,
    linkage="ward"
)

cluster_labels = model.fit_predict(X_scaled)

print("Number of clusters:", n_clusters)
print("Clustering completed successfully.")
```

## 10. Add Cluster Labels

```python
df["Cluster"] = cluster_labels

display_columns = [
    "CustomerID",
    "Age",
    "Annual Income (k$)",
    "Spending Score (1-100)",
    "Cluster"
]

display_columns = [col for col in display_columns if col in df.columns]

display(df[display_columns].head(15))
```

## 11. Customers in Each Cluster

```python
cluster_counts = df["Cluster"].value_counts().sort_index()

print(cluster_counts)
```

## 12. Visualize Customer Segments

```python
plt.figure(figsize=(10, 7))

for cluster in sorted(df["Cluster"].unique()):
    cluster_data = df[df["Cluster"] == cluster]

    plt.scatter(
        cluster_data["Annual Income (k$)"],
        cluster_data["Spending Score (1-100)"],
        label=f"Cluster {cluster}",
        alpha=0.7
    )

plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.title("Hierarchical Clustering - Customer Segmentation")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## 13. Cluster Summary

```python
cluster_summary = (
    df.groupby("Cluster")[
        ["Age", "Annual Income (k$)", "Spending Score (1-100)"]
    ]
    .mean()
    .round(2)
)

display(cluster_summary)
```

## 14. Cluster Size

```python
cluster_size = (
    df.groupby("Cluster")
    .size()
    .reset_index(name="Customer Count")
)

display(cluster_size)
```

## 15. Customer Segment Interpretation

The cluster averages can be used to understand the customer groups:

- **High Income + High Spending:** potential high-value customers
- **High Income + Low Spending:** potential growth opportunity
- **Low Income + High Spending:** strong spending behavior
- **Low Income + Low Spending:** lower-spending segment

The exact interpretation should be based on the generated cluster averages.


## 16. K-Means vs Hierarchical Clustering

```python
comparison = pd.DataFrame({
    "K-Means": [
        "Centroid-based",
        "Requires K",
        "Iterative",
        "Uses centroids"
    ],
    "Hierarchical": [
        "Hierarchy-based",
        "Can inspect dendrogram",
        "Builds hierarchy",
        "Uses cluster merging"
    ]
}, index=[
    "Approach",
    "Number of Clusters",
    "Process",
    "Main Concept"
])

display(comparison)
```

## 17. Save the Clustered Dataset

```python
output_file = "Mall_Customers_Hierarchical_Clustered.csv"

df.to_csv(output_file, index=False)

print(f"Clustered dataset saved as: {output_file}")
```

The project successfully applied hierarchical clustering to customer data.

### Key Learnings

- Hierarchical clustering groups similar observations by building a hierarchy.
- A dendrogram provides a visual representation of cluster merging.
- Feature scaling is important when features have different scales.
- Agglomerative clustering starts with individual observations and progressively merges them.
- Customer segmentation can reveal useful behavioral patterns without predefined labels.

### Conclusion

Hierarchical Clustering is useful when we want to understand relationships between observations and visually inspect possible cluster structures before selecting a final number of clusters.
